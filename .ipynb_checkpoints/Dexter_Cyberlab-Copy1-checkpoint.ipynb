```python

#Time-Based Features
df['EngagementRate'] = df['TimeSpentOnCourse'] / (df['CompletionRate'] + 1e-6)
df['QuizScorePerAttempt'] = df['QuizScores'] / (df['NumberOfQuizzesTaken'] + 1e-6)
```

```python

df['LowEffortUser'] = ((df['TimeSpentOnCourse'] < 10) & (df['CompletionRate'] < 20)).astype(int)

```

```python
#Engagement Segmentation
def categorize_time_spent(time):
    if time < 10:
        return "Low"
    elif time < 50:
        return "Medium"
    else:
        return "High"

df['TimeCategory'] = df['TimeSpentOnCourse'].apply(categorize_time_spent)
df['QuizPerformanceCategory'] = pd.cut(df['QuizScores'], bins=[0, 50, 80, 100], labels=["Low", "Medium", "High"])
```

```python
matrix=df.select_dtypes("number").columns
corr = df[matrix].corr().round(3)
top_features = corr[["CourseCompletion"]].abs().sort_values(by="CourseCompletion", ascending=False)
plt.figure(figsize=(8, 5))
sns.heatmap(corr[top_features.index].loc[top_features.index], annot=True, cmap='coolwarm')
plt.title("Top Correlated Features with CourseCompletion after cleaning")
plt.show()
```

```python
from sklearn.model_selection import train_test_split,  GridSearchCV
from sklearn.preprocessing import OneHotEncoder, LabelEncoder,StandardScaler
from sklearn.metrics import classification_report, accuracy_score
from sklearn.impute import SimpleImputer
from sklearn.feature_selection import RFE
import xgboost as xgb
import joblib
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
# Drop UserID
df = df.drop(columns=["UserID"])

# Encode categorical features with LabelEncoder for RFE step
label_encoders = {}
categorical_cols = ["CourseCategory", "TimeCategory", "QuizPerformanceCategory"]
for col in categorical_cols:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    label_encoders[col] = le  # Store encoder for inverse transform if needed

# Define features and target
X = df.drop(columns=["CourseCompletion"])  # Features
y = df["CourseCompletion"]  # Target

# Initialize RandomForest for RFE
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)

# Apply RFE to select top 7 features
rfe = RFE(estimator=rf_model, n_features_to_select=10)
X_rfe = rfe.fit_transform(X, y)
```

```python
# Get selected features
selected_features = X.columns[rfe.support_]
print("Selected Features:", selected_features)

# Split the dataset based on selected features
X_selected = df[selected_features]  # Use only selected features
X_train, X_test, y_train, y_test = train_test_split(X_selected, y, random_state=42, test_size=0.2)
```

```python
categorical = ["CourseCategory"]
numerical = list(set(selected_features) - set(categorical)) 
```

```python
num_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("scaler", StandardScaler())
])

cat_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])
```

```python
preprocessor = ColumnTransformer([
    ("num", num_transformer, numerical),
    ("cat", cat_transformer, categorical)
])
```

```python
rfc_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("classifier", RandomForestClassifier(class_weight="balanced_subsample", random_state=42))
])
```

```python
param_grid = {
    "classifier__n_estimators": [100, 150, 200],  
    "classifier__max_depth": [5, 10, 15],  
    "classifier__min_samples_split": [5, 10],  
    "classifier__min_samples_leaf": [1, 2],  
    "classifier__max_features": ["sqrt", "log2"],  
    "classifier__bootstrap": [True, False]  
}


```

```python
model = GridSearchCV(
    rfc_pipeline,
    param_grid=param_grid,
    cv=5,
    n_jobs=-1,
    verbose=1
    
)
```

```python
model.fit(X_train, y_train)
```

```python
# extract the best model and evaluate
best_rfc_model = model.best_estimator_
```

```python
#Getting prediction probabilities
y_probs = best_rfc_model.predict_proba(X_test)[:, 1]  # Probabilities for class 1

# Setting a lower threshold 
threshold = 0.35
y_preds_adjusted = (y_probs >= threshold).astype(int)

# Evaluate the new threshold
print("New Classification Report:\n", classification_report(y_test, y_preds_adjusted))
```

```python
# Extract the trained XGBoost model from the pipeline
rfc_model = best_rfc_model.named_steps["model"]

# Get feature importance values
feature_importance = rfc_model.feature_importances_

# Retrieve feature names after preprocessing
categorical_features = ["protocol_type", "encryption_used", "browser_type"]
numerical_features = [
    "network_packet_size", "ip_reputation_score", "failed_logins",
    "unusual_time_access", "session_duration_log", "login_attempts_log"
]

one_hot_encoder = best_rfc_model.named_steps["preprocessor"].named_transformers_["cat"]
one_hot_feature_names = one_hot_encoder.get_feature_names_out(categorical_features)

final_feature_names = np.concatenate([numerical_features, one_hot_feature_names])

# Create a DataFrame for feature importance
feat_imp_df = pd.DataFrame({"Feature": final_feature_names, "Importance": feature_importance})

# Get the most important feature
most_important_features = feat_imp_df.sort_values(by="Importance", ascending=False).head(5)

print("Most Import Features:", most_important_features)
```
